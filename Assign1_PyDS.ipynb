```python
#Current Job data set
job_data = { "Job_ID": [1, 2, 3, 4, 5],
"Role": ["Data Analyst", "Data Scientist", "Business Analyst", "Data Engineer", "MLEngineer"],
"Company": ["ABC Corp", "XyZ Ltd", "DataWorks", "InfraTech", "AI Labs"],
"Skills": [
"Excel SQL Python",
"Python ML Statistics",
"Excel SQL PowerBI",
"Python SQL Spark",
"Python ML DeepLearning"
]}


# add a new job postings in exising data set
def append_job_data(Role, Company, Skills):
  job_data["Job_ID"].append(len(job_data["Job_ID"])+1)
  job_data["Role"].append(Role)
  job_data["Company"].append(Company)
  job_data["Skills"].append(Skills)
  return job_data

  #Modify the skills array
def clean_skills(skill):
  skill = skill.lower()
  skill = skill.replace(", ", " ")
  skill = " ".join(skill.split())
  return skill

#get the data from skill
def find_jobs_by_skill(skill_name):
  import csv

  skill_jobs = []

  with open("job_skills.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Job_ID", "Role", "Company", "Skills"])

  for job_id, role, company, skills in zip(job_data["Job_ID"],job_data["Role"],job_data["Company"], job_data["Skills"]):
      if skill_name in skills.lower().strip():
        skill_jobs.append(job_id)
        skill_jobs.append(role)
        skill_jobs.append(company)
        skill_jobs.append(skills)

  return skill_jobs
  writer.writerow(skill_jobs)

#get the input from user
i=0
Add_data=int(input("How many job posting till need to add in job data"))
while i < Add_data:
  Role = input("enter role")
  Company = input("enter company")
  Skills= input("enter skills")
  updatedlist=append_job_data(Role,Company,Skills)
  i+=1
  if i==Add_data:
    break

# call the function to modify the skills
job_data["Skills"] = [
clean_skills(skill)
for skill in job_data["Skills"]
]

#split the skills into single string
skill_set=[]
for skill in job_data["Skills"]:
  skill_set.extend(skill.split())

#get the count of total skills
skill_count = {}
for skill in skill_set:
    skill_count[skill] = skill_count.get(skill, 0) + 1

#Top 5 skills
top_skills = sorted(
    skill_count.items(),
    key=lambda x: x[1],
    reverse=True
)

# Get the uniques skills from job_data & with ascending order
skill_unique = set(skill_set)
asc_skill_unique = sorted(skill_unique)

#print the statements
print(f"Current list of job_data {updatedlist}")
print(f"extrack the skills array {job_data["Skills"]}")
print(f"Normal list of skills after splitted {skill_set}")
print(f"Count of skills based on skill group {skill_count}")

print("Top skills by frequency:")

for skill, count in top_skills[:5]:
    print(f"{skill}: {count}")

print(f"No of Roles in job_data: {len(job_data["Role"])}")
print(f"Unique skills in job_data: {skill_unique}")
print(f"Unique skills in job_data with alphabatical order: {asc_skill_unique}")

skill_jobs = []
skill_name=input("enter skill name to get the Role in job data")
for role, skills in zip(job_data["Role"], job_data["Skills"]):
    if skill_name in skills.lower():
        skill_jobs.append(role)

print(f"Jobs mentioning {skill_name}:", skill_jobs)

skill_name=input("enter skill name to get the Job_ID,Role,Company in job data")
find_jobs=find_jobs_by_skill(skill_name)
print(f"Data matched with skill provided: {find_jobs}")
```
